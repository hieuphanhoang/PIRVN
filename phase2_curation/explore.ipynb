PIRVN — Exploratory Data Analysis

```python
import json
import matplotlib.pyplot as plt
import numpy as np
from collections import Counter
from pathlib import Path
```

```python
data_dir = Path("curated_data")

with open(data_dir / "train.json", encoding="utf-8") as f:
    train = json.load(f)

with open(data_dir / "val.json", encoding="utf-8") as f:
    val = json.load(f)

with open(data_dir / "test.json", encoding="utf-8") as f:
    test = json.load(f)

all_items = train + val + test

print(f"Train : {len(train):,} items")
print(f"Val   : {len(val):,} items")
print(f"Test  : {len(test):,} items")
print(f"Total : {len(all_items):,} items")
```

```python
prices = [item["price"] for item in all_items]
lengths = [len(item.get("full", "")) for item in all_items]

print("=== Price Stats (VND) ===")
print(f"  Average : {np.mean(prices):,.0f}")
print(f"  Median  : {np.median(prices):,.0f}")
print(f"  Min     : {np.min(prices):,.0f}")
print(f"  Max     : {np.max(prices):,.0f}")

print()
print("=== Description Length Stats (chars) ===")
print(f"  Average : {np.mean(lengths):.1f}")
print(f"  Median  : {np.median(lengths):.1f}")
print(f"  Min     : {np.min(lengths)}")
print(f"  Max     : {np.max(lengths)}")
```

Price Distribution

```python
plt.figure(figsize=(15, 6))
bins = range(500_000, 50_000_000 + 1_000_000, 1_000_000)
plt.hist(prices, bins=bins, color="blueviolet", edgecolor="white", linewidth=0.3)
plt.xlabel("Price (VND)")
plt.ylabel("Count")
plt.title("Price Distribution")
ax = plt.gca()
ax.xaxis.set_major_formatter(
    plt.FuncFormatter(lambda x, _: f"{x/1_000_000:.0f}M")
)
plt.tight_layout()
plt.show()
```

Description Length Distribution

```python
plt.figure(figsize=(15, 6))
bins = range(0, 4000 + 100, 100)
plt.hist(lengths, bins=bins, color="skyblue", edgecolor="white", linewidth=0.3)
plt.xlabel("Description Length (chars)")
plt.ylabel("Count")
plt.title("Description Length Distribution")
plt.tight_layout()
plt.show()
```

Category Distribution

```python
categories = Counter(item.get("category", "Unknown") for item in all_items)
cat_labels = [k for k, _ in categories.most_common()]
cat_counts = [v for _, v in categories.most_common()]

plt.figure(figsize=(15, 6))
bars = plt.bar(cat_labels, cat_counts, color="goldenrod", edgecolor="white")
for bar, count in zip(bars, cat_counts):
    plt.text(
        bar.get_x() + bar.get_width() / 2,
        bar.get_height() + max(cat_counts) * 0.005,
        f"{count:,}",
        ha="center",
        va="bottom",
        fontsize=9,
    )
plt.xlabel("Category")
plt.ylabel("Count")
plt.title("Category Distribution")
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

Source Distribution

```python
sources = Counter(item.get("source", "Unknown") for item in all_items)
src_labels = list(sources.keys())
src_counts = list(sources.values())

plt.figure(figsize=(15, 6))
plt.pie(
    src_counts,
    labels=src_labels,
    autopct="%1.1f%%",
    startangle=140,
)
plt.title("Items by Source Site")
plt.tight_layout()
plt.show()
```

Price by Category

```python
from collections import defaultdict

prices_by_cat = defaultdict(list)
for item in all_items:
    cat = item.get("category", "Unknown")
    prices_by_cat[cat].append(item["price"])

sorted_cats = sorted(prices_by_cat.keys())
box_data = [prices_by_cat[c] for c in sorted_cats]

plt.figure(figsize=(15, 6))
plt.boxplot(box_data, labels=sorted_cats, patch_artist=True)
plt.xlabel("Category")
plt.ylabel("Price (VND)")
plt.title("Price by Category")
ax = plt.gca()
ax.yaxis.set_major_formatter(
    plt.FuncFormatter(lambda y, _: f"{y/1_000_000:.0f}M")
)
plt.xticks(rotation=45, ha="right")
plt.tight_layout()
plt.show()
```

Price vs Description Length

```python
plt.figure(figsize=(15, 6))
plt.scatter(lengths, prices, s=0.5, color="red", alpha=0.6)
plt.xlabel("Description Length (chars)")
plt.ylabel("Price (VND)")
plt.title("Price vs Description Length")
ax = plt.gca()
ax.yaxis.set_major_formatter(
    plt.FuncFormatter(lambda y, _: f"{y/1_000_000:.0f}M")
)
plt.tight_layout()
plt.show()
```
